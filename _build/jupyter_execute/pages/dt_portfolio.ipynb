# ECMWF's Destination Earth Data Porfolio

This section refers only to Digital Twin Data, which is the data produced by ECMWF. ECMWF has currently two high priority digital twins, the Climate-Dt and the Extremes DT.


## Climate Digital Twin

Historical simulations (CMIP6) from the Climate DT are currently available from ca 1990 untill ca 2007. 

Projections are available from ca
