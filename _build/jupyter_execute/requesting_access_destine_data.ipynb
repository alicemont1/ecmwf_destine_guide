# Requesting Access to Destination Earth Data

To access the Digital Twin datasets, you need to first create an account on the **Destination Earth Platform** at [https://platform.destine.eu/](https://platform.destine.eu/). Once your account is created, you can apply for **upgraded access**. Please note that upgraded access is currently available only to specific user groups, which include:

## Eligible User Groups:

- **Contractors of the 3Es (ESA/ECMWF/EUMETSAT)**: Individuals contributing to the implementation and further development of **DestinE**.
- **Staff of the 3Es (ESA/ECMWF/EUMETSAT)**.
- **EU Institutions, Agencies, and Bodies**: Public authorities of EU Member States and Digital Europe Programme (DEP) Associated Countries.
- **Selected Third-Country Public Authorities**: Aligned with the EU’s international commitments.
- **SMEs and Start-ups**: Based in EU Member States or DEP-associated countries, with their domain of activity and usage consistent with the objectives of **DestinE**.
- **Public Academia and Research Teams**: From EU Member States or DEP-associated countries, provided their research aligns with the goals of **DestinE**.
- **Third-Country Academia and Research Teams, and SMEs**: On a case-by-case basis, for entities whose activities support **DestinE** objectives and do not compromise the EU’s strategic assets, autonomy, or security.

Once your account is created and you’ve selected the appropriate user category, you can apply for upgraded access at: [Access Policy Upgrade](https://platform.destine.eu/access-policy-upgrade/).

Your request will then be reviewed and approved. For more details, please refer to the [Access Policy Upgrade page](https://platform.destine.eu/access-policy-upgrade/).
