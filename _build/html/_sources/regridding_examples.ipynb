Client side regridding with Earthkit Data

```python
import earthkit.regrid
```

```python
request = {
    'activity': 'ScenarioMIP',
    'class': 'd1',
    'dataset': 'climate-dt',
    'date': '20200102',
    'experiment': 'SSP3-7.0',
    'expver': '0001',
    'generation': '1',
    'levtype': 'sfc',
    'model': 'IFS-NEMO',
    'param': '134',
    'realization': '1',
    'resolution': 'standard',
    'stream': 'clte',
    'time': '0100', # '0100/0200/0300/0400/0500/0600'
    'type': 'fc'
}
```

```python
data = earthkit.data.from_source("polytope", "destination-earth", request, address="polytope.lumi.apps.dte.destination-earth.eu", stream=False)
```

```python
data.ls() # healpix grid
```

```python
# regrid to 1x1 degree
out_grid = {"grid": [1,1]}
data_interpolated = earthkit.regrid.interpolate(data, out_grid=out_grid, method="linear")
```

```python
data_interpolated.ls() # now on regular lat lon grid
```

```python
import earthkit.data

request_original = {
        "class": "d1",
        "expver": "0001",
        "stream": "oper",
        "dataset": "extremes-dt",
        "date": "-14",
        "time": "0000",
        "type": "fc",
        "levtype": "sfc",
        "step": "13",
        "param": "167",
        # "feature" : {
        # "type" : "boundingbox",
        # "points" : [[30, -15], [75, 42]],
        # "axes" : ["latitude", "longitude"]
        # },
        "area": "75/-15/30/42.5",
        "grid": "0.01/0.01"
    }
original_grid = earthkit.data.from_source("polytope", "destination-earth", request_original, address="polytope.lumi.apps.dte.destination-earth.eu", stream=False)
```

```python
data = original_grid.to_xarray()
data['2t']
```

```python
data['2t'].max() - 273.15
```

```python
import earthkit.plots

chart = earthkit.plots.Map(domain=["Spain"])

chart.grid_cells(data['2t'],
    x="longitude",
    y="latitude",
    units="celsius", levels=range(0, 40, 5), colors="bwr",
)

chart.title()
chart.legend()
chart.cities()
chart.gridlines()
chart.coastlines()
chart.ocean(color="black")

chart.show()
```
